```julia
sites = siteinds(state.mps)

i, j = 3, 10   # choose sites

ai   = op("a",  sites, i)    # a_i
adag = op("a†", sites, j)  # a†_j

psi2 = copy(state.mps)
psi2[i] = ai   * psi2[i]
psi2[j] = adag * psi2[j]

val = loginner(state.mps, psi2) - loginner(state.mps, state.mps)   # ⟨a_i a†_j⟩
```
