```julia
using Plots
```

```julia
CMI(Ra, Rb, Rc) = 1/2 * log(log(Rc/Ra) / log(Rb/Ra))

approx_CMI(Ra, Rb, Rc) = 1/2 * (Rc - Rb) / ((Rb - Ra) * log(Rb/Ra))
```

```julia
plt = plot()
for Rc in [1e2, 1e3, 1e4]
    Ra = 1
    Rbs = 2:Rc
    plt = plot!(plt, collect(Rbs)/Rc, CMI.(Ra, Rbs, Rc), label="Rc = $Rc", xlabel="Rb/Rc", ylabel="CMI")
end

for Rc in [1e2, 1e3, 1e4]
    Ra = 1
    Rbs = Rc÷10:Rc
    plt = plot!(plt, collect(Rbs)/Rc, approx_CMI.(Ra, Rbs, Rc), label="Rc = $Rc", xlabel="Rb/Rc", ylabel="CMI", linestyle=:dash)
end

plt
```
